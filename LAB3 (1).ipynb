```python
from scipy.stats import pearsonr

EntradaCampus = Entrada_H1 + Entrada_M1 + Entrada_I1
SalidaCampus = Salida_H2 + Salida_M2 + Salida_I2

EntradaBiblio = Entrada_HBI + Entrada_MBI + Entrada_IBI + Salida_HBI + Salida_MBI + Salida_IBI

EntradaMu = Entrada_HMU + Entrada_MMU + Entrada_IMU + Salida_HMU + Salida_MMU + Salida_IMU

fig, axes = plt.subplots(2, 3, figsize=(15, 8))

#Grafica 1
r, p_value = pearsonr(EntradaCampus, SalidaCampus)

h1 = axes[0, 0].hist2d(EntradaCampus, SalidaCampus, bins=4, cmap="viridis")
axes[0, 0].set_title(f"CoeficientePearson:{r:.2f}")
axes[0, 0].set_xlabel("Entrada Campus")
axes[0, 0].set_ylabel("Salida Campus")
fig.colorbar(h1[3], ax=axes[0, 0])

print("\nr =", r)
print("p-value =", p_value)


#Grafica 2
r, p_value = pearsonr(EntradaCampus, EntradaBiblio)

h2 = axes[0, 1].hist2d(EntradaCampus, EntradaBiblio, bins=4, cmap="viridis")
axes[0, 1].set_title(f"CoeficientePearson:{r:.2f}")
axes[0, 1].set_xlabel("Entrada Campus")
axes[0, 1].set_ylabel("Entrada Biblioteca")
fig.colorbar(h2[3], ax=axes[0, 1])

print("\nr =", r)
print("p-value =", p_value)


#Grafica 3
r, p_value = pearsonr(EntradaCampus, EntradaMu)

h3 = axes[0, 2].hist2d(EntradaCampus, EntradaMu, bins=4, cmap="viridis")
axes[0, 2].set_title(f"CoeficientePearson:{r:.2f}")
axes[0, 2].set_xlabel("Entrada Campus")
axes[0, 2].set_ylabel("Entrada Museo")
fig.colorbar(h3[3], ax=axes[0, 2])

print("\nr =", r)
print("p-value =", p_value)


#Grafica 4
r, p_value = pearsonr(Entrada_H1, Entrada_M1)

h4 = axes[1, 0].hist2d(Entrada_H1, Entrada_M1, bins=4, cmap="viridis")
axes[1, 0].set_title(f"CoeficientePearson:{r:.2f}")
axes[1, 0].set_xlabel("Hombres Campus")
axes[1, 0].set_ylabel("Mujeres Campus")
fig.colorbar(h4[3], ax=axes[1, 0])

print("\nr =", r)
print("p-value =", p_value)



#Grafica 5
r, p_value = pearsonr(Entrada_H1, Entrada_I1)

h5 = axes[1, 1].hist2d(Entrada_H1, Entrada_I1, bins=4, cmap="viridis")
axes[1, 1].set_title(f"CoeficientePearson:{r:.2f}")
axes[1, 1].set_xlabel("Hombres Campus")
axes[1, 1].set_ylabel("Inter Campus")
fig.colorbar(h5[3], ax=axes[1, 1])

print("\nr =", r)
print("p-value =", p_value)


#Grafica 6
r, p_value = pearsonr(Entrada_I1, Entrada_M1)

h6 = axes[1, 2].hist2d(Entrada_I1, Entrada_M1, bins=4, cmap="viridis")
axes[1, 2].set_title(f"CoeficientePearson:{r:.2f}")
axes[1, 2].set_xlabel("Inter Campus")
axes[1, 2].set_ylabel("Mujeres Campus")
fig.colorbar(h6[3], ax=axes[1, 2])

print("\nr =", r)
print("p-value =", p_value)


plt.tight_layout()
plt.show()
```

```python
# 1. Promedio y Varianza de Entrada_M1
average_value_M1 = np.mean(Entrada_M1)
variance_value_M1 = np.var(Entrada_M1)

print(f"El promedio de Entrada_M1 es: {average_value_M1:.2f}")
print(f"La varianza de Entrada_M1 es: {variance_value_M1:.2f}")

# Generar datos del histograma para Entrada_M1
min_val_M1 = Entrada_M1.min()
max_val_M1 = Entrada_M1.max()
bins_M1 = np.arange(min_val_M1, max_val_M1 + 2) - 0.5 # Centra los enteros
counts_M1, bin_edges_M1 = np.histogram(Entrada_M1, bins=bins_M1)
bin_centers_M1 = np.arange(min_val_M1, max_val_M1 + 1)

print("\nCounts (frecuencia de cada valor) para Entrada_M1:", counts_M1)
print("Bin Centers (valores correspondientes) para Entrada_M1:", bin_centers_M1)

# 2. Valor más probable (Moda) de Entrada_M1
most_probable_index_M1 = np.argmax(counts_M1)
most_probable_value_M1 = bin_centers_M1[most_probable_index_M1]

print(f"\nEl valor más probable (moda) de Entrada_M1 es: {most_probable_value_M1}")

# 3. Valor que divide el histograma en dos partes de áreas iguales (Mediana) de Entrada_M1
median_value_M1 = np.median(Entrada_M1)

print(f"\nEl valor que divide el histograma en dos partes de áreas iguales (mediana) de Entrada_M1 es: {median_value_M1}")

# 4. Número de máximos locales para Entrada_M1
local_maxima_count_M1 = 0
if len(counts_M1) > 1:
    if counts_M1[0] > counts_M1[1]:
        local_maxima_count_M1 += 1
    for i in range(1, len(counts_M1) - 1):
        if counts_M1[i] > counts_M1[i-1] and counts_M1[i] > counts_M1[i+1]:
            local_maxima_count_M1 += 1
    if counts_M1[-1] > counts_M1[-2]:
        local_maxima_count_M1 += 1

print(f"\nLa variable Entrada_M1 presenta {local_maxima_count_M1} máximos locales.")

# 5. Valor menos probable pero distinto de cero para Entrada_M1
non_zero_counts_M1 = counts_M1[counts_M1 > 0]
non_zero_counts_indices_M1 = np.where(counts_M1 > 0)[0]

if len(non_zero_counts_M1) > 0:
    min_freq_non_zero_M1 = np.min(non_zero_counts_M1)
    min_freq_indices_M1 = non_zero_counts_indices_M1[non_zero_counts_M1 == min_freq_non_zero_M1]
    least_probable_non_zero_values_M1 = bin_centers_M1[min_freq_indices_M1]
    print(f"\nEl valor(es) menos probable(s) pero distinto(s) de cero de Entrada_M1 es/son: {least_probable_non_zero_values_M1}")
else:
    print("\nNo hay valores distintos de cero en el histograma de Entrada_M1.")
```
