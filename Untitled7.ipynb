```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar el diseño
sns.set_theme(style="whitegrid")
plt.figure(figsize=(8, 5))

# Creamos el histograma usando el nombre exacto de tu columna
sns.histplot(data=df, x="Temperature (K)", kde=True, color="purple", bins=20)

# Títulos profesionales para tu reporte
plt.title("Distribución de la Temperatura Absoluta (Variable X)", fontsize=14, fontweight='bold')
plt.xlabel("Temperatura (Kelvin)", fontsize=12)
plt.ylabel("Cantidad de Estrellas", fontsize=12)

# Mostrar la gráfica
plt.show()
```

```python
import numpy as np
import scipy.stats as stats

# 1. Obtener los datos básicos de la muestra
n = len(df)
media_X = df["Temperature (K)"].mean()
std_X = df["Temperature (K)"].std()

# 2. Definir el valor de la hipótesis nula (mu_0)
mu_0 = 10000

# 3. Calcular el estadístico de prueba (t-calculado)
t_stat = (media_X - mu_0) / (std_X / np.sqrt(n))

# 4. Calcular el p-valor (para una prueba bilateral)
p_valor = 2 * (1 - stats.t.cdf(abs(t_stat), df=n-1))

print(f"Tamaño de la muestra (n): {n}")
print(f"Media muestral (X_bar): {media_X:.2f} K")
print(f"Desviación estándar muestral (s): {std_X:.2f} K")
print(f"Estadístico t calculado: {t_stat:.4f}")
print(f"P-valor: {p_valor:.4f}")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Definimos formalmente nuestras variables X e Y
X_var = "Temperature (K)"
Y_var = "Absolute magnitude(Mv)"

# =========================================================
# PUNTO 5: Graficar el Histograma 2D
# =========================================================
plt.figure(figsize=(9, 6))
# 'cbar=True' añade la barra lateral que explica la intensidad del color
sns.histplot(data=df, x=X_var, y=Y_var, cbar=True, cmap="magma", bins=20)

plt.title("Punto 5: Histograma 2D entre Temperatura (X) y Magnitud Absoluta (Y)", fontsize=12, fontweight='bold')
plt.xlabel("Temperatura (Kelvin) - Variable X", fontsize=11)
plt.ylabel("Magnitud Absoluta (Mv) - Variable Y", fontsize=11)
plt.gca().invert_yaxis() # En astronomía, las magnitudes menores/negativas son más brillantes, invertimos para respetar el estándar físico (Diagrama HR)
plt.show()

# =========================================================
# PUNTOS 6, 7 y 8: Cálculos de mu_Y, Y_up e Y_down
# =========================================================
# 6. Promedio de Y
mu_Y = df[Y_var].mean()

# Buscamos qué proporción de los datos totales están por debajo del promedio mu_Y
prop_menor_mu = (df[Y_var] <= mu_Y).mean()

# 7 y 8. Usamos cuantiles empíricos para asegurar exactamente el 25% a cada lado
Y_up = df[Y_var].quantile(prop_menor_mu + 0.25)
Y_down = df[Y_var].quantile(prop_menor_mu - 0.25)

print("--- RESULTADOS NUMÉRICOS ---")
print(f"Punto 6: Promedio de Magnitud Absoluta (mu_Y) = {mu_Y:.4f}")
print(f"Punto 7: Valor límite superior (Y_up) = {Y_up:.4f}")
print(f"Punto 8: Valor límite inferior (Y_down) = {Y_down:.4f}")

# =========================================================
# PUNTO 9: Construcción de la base de datos depurada
# =========================================================
df_depurada = df[(df[Y_var] >= Y_down) & (df[Y_var] <= Y_up)]

print("\n--- COMPROBACIÓN DEL PUNTO 9 ---")
print(f"Cantidad de datos originales: {len(df)}")
print(f"Cantidad de datos en la base depurada: {len(df_depurada)} (Debe ser exactamente {len(df)//2})")
```

```python
# =========================================================
# FASE FINAL: Comparación de Correlaciones y Gráficos
# =========================================================

# 计算皮尔逊相关系数 (Calcular coeficientes de Pearson)
r_original = df[X_var].corr(df[Y_var])
r_depurada = df_depurada[X_var].corr(df_depurada[Y_var])

print("--- ANÁLISIS DE CORRELACIÓN ---")
print(f"Coeficiente 'r' de Pearson (Datos Originales): {r_original:.4f}")
print(f"Coeficiente 'r' de Pearson (Datos Depurados): {r_depurada:.4f}")

# Graficar la comparación lado a lado
fig, axes = plt.subplots(1, 2, figsize=(14, 5), sharey=True)

# Gráfica 1: Datos Originales
sns.scatterplot(data=df, x=X_var, y=Y_var, ax=axes[0], color="blue", alpha=0.6)
sns.regplot(data=df, x=X_var, y=Y_var, ax=axes[0], scatter=False, color="red")
axes[0].set_title(f"Datos Originales (240 estrellas)\nCon correlación r = {r_original:.3f}")
axes[0].set_xlabel("Temperatura (K)")
axes[0].set_ylabel("Magnitud Absoluta (Mv)")
axes[0].invert_yaxis()

# Gráfica 2: Datos Depurados
sns.scatterplot(data=df_depurada, x=X_var, y=Y_var, ax=axes[1], color="green", alpha=0.6)
sns.regplot(data=df_depurada, x=X_var, y=Y_var, ax=axes[1], scatter=False, color="red")
axes[1].set_title(f"Datos Depurados (120 estrellas centrales)\nCon correlación r = {r_depurada:.3f}")
axes[1].set_xlabel("Temperatura (K)")
axes[1].invert_yaxis()

plt.tight_layout()
plt.savefig("comparativa_correlacion.png", dpi=300)
plt.show()
```
